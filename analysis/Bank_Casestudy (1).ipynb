```python
def uni_target_categorical_analysis(column):
    plt.figure(figsize=(25,10))
    plt.subplot(1,2,1)
    ax1=sns.countplot(x="TARGET",hue=column,data=df_target_non_default)
    plt.title(f'{column} CONTRIBUTION OF NON-DEFAULT')
    plt.subplot(1,2,2)
    ax2=sns.countplot(x="TARGET",hue=column,data=df_target_default)
    plt.title(f'{column} CONTRIBUTION OF DEFAULT')

    total_count=len(df_target_non_default)
    for p in ax1.patches:
        percentage=f'{100*p.get_height()/total_count:.3}%'
        ax1.annotate(percentage,
                (p.get_x()+p.get_width()/2,p.get_height()),
                ha="center",va="bottom",color="black",fontsize=10,weight="bold")
    
    total_count1=len(df_target_default)
    for p in ax2.patches:
        percentage=f'{100*p.get_height()/total_count1:.3}%'
        ax2.annotate(percentage,
                (p.get_x()+p.get_width()/2,p.get_height()),
                ha="center",va="bottom",color="black",fontsize=10,weight="bold")


    plt.show()
```
